```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import OrdinalEncoder

# 1. Data
data = [
    ['>=9','yes','very good','good','yes'],
    ['>=9','yes','good','good','yes'],
    ['>=9','no','very good','moderate','yes'],
    ['>=8','yes','good','good','yes'],
    ['>=8','yes','very good','moderate','yes'],
    ['>=8','no','good','moderate','no'],
    ['>=8','yes','avg','good','yes'],
    ['>=8','no','avg','poor','no'],
    ['>=8','yes','good','moderate','yes'],
    ['>=8','no','good','good','no'],
    ['<8','yes','good','moderate','no'],
    ['<8','no','avg','poor','no'],
    ['<8','yes','avg','moderate','no'],
    ['<8','no','avg','poor','no'],
    ['<8','yes','good','good','yes'],
    ['<8','no','avg','moderate','no'],
    ['>=9','yes','very good','good','yes'],
    ['>=8','no','very good','moderate','yes'],
    ['<8','yes','avg','good','no'],
    ['<8','no','avg','poor','no'],
]
cols = ['CGPA','Interactiveness','PracticalKnowledge','Communication','JobOffer']
df = pd.DataFrame(data, columns=cols)

# 2. Encode features
enc = OrdinalEncoder()
X = enc.fit_transform(df[cols[:-1]])
y = df['JobOffer'].map({'no': 0, 'yes': 1})

# 3. Train Naive Bayes
model = CategoricalNB(alpha=1.0)
model.fit(X, y)

# 4. Predict for a given instance
instance = [['>=8', 'yes', 'good', 'moderate']]
proba = model.predict_proba(enc.transform(instance))[0]
print(f"P(no)={proba[0]:.4f}   P(yes)={proba[1]:.4f}")

# 5. Bar graph
plt.bar(['No', 'Yes'], proba, color=['red', 'green'])
plt.ylabel('Probability')
plt.title('Job Offer Likelihood')
for i, v in enumerate(proba):
    plt.text(i, v + 0.02, f"{v:.2%}", ha='center')
plt.ylim(0, 1)
plt.show()
```

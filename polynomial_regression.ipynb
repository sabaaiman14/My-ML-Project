```python

import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

#Dataset
x = np.array([-2,-1,0,1,2]).reshape(-1,1)
y = np.array([9,2,1,6,17])

#Polynomial features
poly = PolynomialFeatures(degree = 2)
X_poly = poly.fit_transform(x)

#Train Model
model = LinearRegression()
model.fit(X_poly, y)

#Predictions
y_pred = model.predict(X_poly)

#Equations
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)

#MSE
mse = mean_squared_error(y,y_pred)

print("Predicted values:", y_pred)
print("Train MSE:", mse)

# Create smooth x values for the curve
x_curve = np.linspace(-2, 2, 100).reshape(-1, 1)

# Create polynomial features for smooth x values
X_curve = poly.transform(x_curve)

# Predict y values for the curve
y_curve = model.predict(X_curve)

# Plot actual data points
plt.scatter(x, y, color="red", label="Actual Data")

# Plot polynomial regression curve
plt.plot(x_curve, y_curve, color="blue", label="Polynomial Regression")

# Labels and title
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Polynomial Regression (Degree 2)")

# Show grid and legend
plt.grid(True)
plt.legend()

# Display graph
plt.show()
```

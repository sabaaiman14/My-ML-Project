```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv("k_fold_linear_regression.csv", delim_whitespace=True)
df.columns = df.columns.str.strip()
print(df)
```

```python
X= "Hours Studied"
y="Marks"
x=df[["Hours_Studied(x)"]]
y=df["Marks(y)"]
print( "x:")
print(x)
print("\ny:")
print(y)
```

```python
kf = KFold(n_splits=5, shuffle=False)
print("number of folds:", kf.n_splits)
```

```python
mse_values=[]

fold=1
for train_index, validation_index in kf.split(x):
  X_train=x.iloc[train_index]

  y_train= y.iloc[train_index]

  X_validation=x.iloc[validation_index]

  y_validation=y.iloc[validation_index]

  model=LinearRegression()
  model.fit(X_train,y_train)
  slope= model.coef_[0]
  intercept=model.intercept_
  y_pred=model.predict(X_validation)
  mse=mean_squared_error(y_validation,y_pred)
  mse_values.append(mse)

  print("======== fold",fold,"=======")
  print("training recrds:",list(train_index))
  print("validation records:",list(validation_index))
  print("slope:",round(slope ,4))
  print("intercept:",round(intercept,4))
  print("regression euation:")
  print("y=",round (intercept,4),"+",round(slope,4),"X")
  print("actual values:",list(y_validation))
  print("predicted values:",[round(value,4)for value in y_pred])
  print("validation MSE:",round(mse,4))
  print()
  fold+=1
```

```python
print("mse for each fold")
for i ,mse in enumerate(mse_values, start=1):
  print("fold",i,"mse:",round(mse,4))
  average_mse=sum(mse_values)/len(mse_values)
print("average cross-validation MSE:",round(average_mse,4))
```
